## Your listening decisions

Choose one main error per item. `no_audio_error` records that you listened but found no audible error; that item is left out of the submission tags.


```python
# Participant's listening decisions; full observations are in reports/task1/listening_notes.csv.
# One primary category per clip; number_error is primary for the hard items.
LABELS = {
    'val_hd_08': 'number_error',
    'val_hd_01': 'number_error',
    'val_hd_03': 'number_error',
    'val_hd_07': 'number_error',
    'val_hg_01': 'english_word_error',
    'val_hd_05': 'number_error',
    'val_hd_06': 'number_error',
    'val_hd_02': 'number_error',
    'val_hg_06': 'english_word_error',
    'val_hg_05': 'english_word_error',
    'val_hg_03': 'english_word_error',
    'val_hd_04': 'number_error',
    'val_hg_04': 'english_word_error',
    'val_hg_02': 'english_word_error',
}
```

```python
allowed = {'skipped_words', 'repeated_words', 'number_error', 'english_word_error', 'speaker_mismatch', 'other', 'no_audio_error'}
with (KIT / 'reports/task1/listening_queue.csv').open(encoding='utf-8', newline='') as f:
    required_ids = {r['item_id'] for r in csv.DictReader(f)}
with (KIT / 'data/val_items.csv').open(encoding='utf-8', newline='') as f:
    valid_ids = {r['id'] for r in csv.DictReader(f)}
pending = sorted(i for i in required_ids if LABELS.get(i) not in allowed)
invalid = {i: tag for i, tag in LABELS.items() if i not in valid_ids or tag not in allowed}
if pending or invalid:
    raise ValueError(f'Complete your listening decisions first. Pending: {pending}; invalid: {invalid}')
tagged = [(i, tag) for i, tag in LABELS.items() if tag != 'no_audio_error']
if not tagged:
    raise ValueError('No audible errors tagged. Recheck the audio before submitting Task 1.')
out = KIT / 'predictions/t1_val_errors.csv'
out.parent.mkdir(parents=True, exist_ok=True)
with out.open('w', encoding='utf-8', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(['item_id', 'category'])
    writer.writerows(tagged)
(KIT / 'reports/task1/listening_decisions.json').write_text(json.dumps(LABELS, indent=2) + '\n', encoding='utf-8')
print(f'Saved {len(tagged)} human-reviewed error tags to {out}')
```
